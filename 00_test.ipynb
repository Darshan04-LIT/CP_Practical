```python
#create a simple if else code to check marks and give
#grade A, B, C, D, or F based on the marks entered by the user
marks = int(input("Enter your marks: "))
if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Your grade is: {grade}")
```
