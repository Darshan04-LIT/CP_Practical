```python
Q-1-e) classifies the conversion using if/elif/else : >= 70% "High, >= 40% "Moderate", otherwise "Low".Test with Kc = 4, Kc=0.5,K=-2.

prompt:classify conversion using if/elif/else: High for conversion ≥ 70%, Moderate for ≥ 40%, otherwise Low. It asks you to test with Kc = 4, Kc = 0.5, and Kc = -2. There isn’t a separate prompt: entry for this part
```

```python
import math


def classify_conversion(conversion_percent):
    if conversion_percent >= 70:
        return "High"
    elif conversion_percent >= 40:
        return "Moderate"
    else:
        return "Low"


for test_kc in (4.0, 0.5, -2.0):
    print(f"Kc = {test_kc}")

    if not math.isfinite(test_kc) or test_kc <= 0:
        print("Invalid Kc: it must be a finite number greater than 0.\n")
        continue

    coefficient_a = test_kc
    coefficient_b = -(3 * test_kc + 1)
    coefficient_c = 2 * test_kc
    discriminant = coefficient_b**2 - 4 * coefficient_a * coefficient_c
    square_root = math.sqrt(discriminant)

    roots = (
        (-coefficient_b + square_root) / (2 * coefficient_a),
        (-coefficient_b - square_root) / (2 * coefficient_a),
    )
    physical_roots = [root for root in roots if 0 <= root <= 1]

    if not physical_roots:
        print("No root is in the physical range 0 <= x <= 1.\n")
        continue

    test_x = physical_roots[0]
    conversion_percent = (test_x / 1.0) * 100
    classification = classify_conversion(conversion_percent)

    print(f"B conversion = {conversion_percent:.3f}%")
    print(f"Classification = {classification}\n")
```

```python
initial_concentration = 1.5  # mol/L
rate_constant = 0.05  # min^-1
time_step = 2  # min
total_time = 20  # min
steps = int(total_time / time_step)

concentration = initial_concentration

print(f"{'Time (min)':>10} {'Concentration (mol/L)':>24} {'Conversion (%)':>16}")
print("-" * 52)

for step in range(steps + 1):
    time = step * time_step
    conversion = (initial_concentration - concentration) / initial_concentration * 100
    print(f"{time:10.0f} {concentration:24.3f} {conversion:16.3f}")

    if step < steps:
        concentration = concentration - rate_constant * concentration * time_step
```

```python
Q-2-3) using a while loop, repeat the simulation untill conversion reaches 50% or time reaches 100min.Print the time needed and the final concentration.

propmt:use a while loop to simulate until conversion reaches 50% or time reaches 100 minutes, then print the time needed and final concentration. 
```

```python
initial_concentration = 1.5  # mol/L
rate_constant = 0.05  # min^-1
time_step = 2  # min
maximum_time = 100  # min

concentration = initial_concentration
time = 0
conversion = 0.0

while conversion < 50 and time < maximum_time:
    concentration = concentration - rate_constant * concentration * time_step
    time += time_step
    conversion = (initial_concentration - concentration) / initial_concentration * 100

print(f"Time needed = {time} min")
print(f"Final concentration = {concentration:.3f} mol/L")
print(f"Final conversion = {conversion:.2f}%")

if conversion < 50:
    print("The 100-minute limit was reached before conversion reached 50%.")
```

```python
Part B.Reactor tags such as REACTOR-07/B-2026 are stored as text.write a program that reads a tag, loops through each character, and prints the number of digits and the sum of digits . show a message for an empty tag and for a tag with no digits .test with REACTOR-07/B-2026, an empty tag and ABCDEF.

prompt:read a reactor tag, loop through its characters, and print the number of digits and their sum
```

```python
def analyze_tag(tag):
    if tag == "":
        print("The tag is empty.")
        return

    digit_count = 0
    digit_sum = 0

    for character in tag:
        if "0" <= character <= "9":
            digit_count += 1
            digit_sum += int(character)

    if digit_count == 0:
        print("The tag contains no digits.")
    else:
        print(f"Number of digits: {digit_count}")
        print(f"Sum of digits: {digit_sum}")


test_tags = ("REACTOR-07/B-2026", "", "ABCDEF")

for tag in test_tags:
    print(f"Tag: {tag!r}")
    analyze_tag(tag)
    print()
```

```python
Q-3.Part A.A laboratory has n candidate catalysts and wants to test r of them together (order does not matter).write a recursive function(n)(remember the base case for 0!)and a function combination(n,r)=n!/(r!(n-r)!) that uses it.Rejects invalid input (negative values,r>n).print the number of possible catalyts sets for n=7, r=2,amd show  the message for n=4, r=6.

PROMPT:write a recursive factorial function with a base case for 0, then use it to calculate combinatination with formula.
```

```python
def recursive_factorial(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("Factorial input must be a non-negative integer.")
    if n == 0:
        return 1
    return n * recursive_factorial(n - 1)


def combination(n, r):
    if not isinstance(n, int) or isinstance(n, bool):
        raise ValueError("n must be an integer.")
    if not isinstance(r, int) or isinstance(r, bool):
        raise ValueError("r must be an integer.")
    if n < 0 or r < 0:
        raise ValueError("n and r must be non-negative.")
    if r > n:
        raise ValueError("r cannot be greater than n.")

    return recursive_factorial(n) // (
        recursive_factorial(r) * recursive_factorial(n - r)
    )


for n, r in ((7, 2), (4, 6)):
    try:
        number_of_sets = combination(n, r)
    except ValueError as error:
        print(f"Invalid input n={n}, r={r}: {error}")
    else:
        print(f"Number of catalyst sets for n={n}, r={r}: {number_of_sets}")
```

```python
Q-3-Part B.During a distillation run, samples are taken at times = fibonacci number * 5min . write an iterative function fib_iter(n) that returns a list of the first n fibonacci numbers, and a recursive function fib_rec(k) that returns the k-th fibonacci tag | sampling time(min).

prompt:Write an iterative function fib_iter(n) that returns the first n Fibonacci numbers and a recursive function fib_rec(k) that returns the kth Fibonacci number. Display each tag and its sampling time in minutes, calculated as the tag multiplied by 5.
```

```python
def fib_iter(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n must be a non-negative integer.")

    fibonacci_numbers = []
    previous_number, current_number = 0, 1

    for _ in range(n):
        fibonacci_numbers.append(previous_number)
        previous_number, current_number = current_number, previous_number + current_number

    return fibonacci_numbers


def fib_rec(k):
    if not isinstance(k, int) or isinstance(k, bool) or k < 0:
        raise ValueError("k must be a non-negative integer.")
    if k == 0:
        return 0
    if k == 1:
        return 1
    return fib_rec(k - 1) + fib_rec(k - 2)


sample_count = 10
fibonacci_tags = fib_iter(sample_count)

print(f"First {sample_count} Fibonacci tags:", fibonacci_tags)
print(f"{'Index':>5} {'Fibonacci tag':>15} {'Sampling time (min)':>21}")
for index, fibonacci_tag in enumerate(fibonacci_tags):
    recursive_tag = fib_rec(index)
    sampling_time = recursive_tag * 5
    print(f"{index:5d} {fibonacci_tag:15d} {sampling_time:21d}")
```

```python
Q-4-d)Carbon dioxide is compressed slowly ina piston-cylinder at constant temperature:

Volume_L            1           2           3           4           5
Pressure_bar        3.02        1.49        1.01        0.76        0.59

Plot pressure vs. volume as a scatter plot.Calculate k as the mean of P*V and draw the reference curve P=k/V on the same graph .Add title,labels with units and legend.
```

```python
import matplotlib.pyplot as plt
import numpy as np

volume_l = np.array([1, 2, 3, 4, 5], dtype=float)
pressure_bar = np.array([3.02, 1.49, 1.01, 0.76, 0.59], dtype=float)

k_bar_l = np.mean(pressure_bar * volume_l)
reference_volume_l = np.linspace(volume_l.min(), volume_l.max(), 200)
reference_pressure_bar = k_bar_l / reference_volume_l

print(f"Mean P*V (k) = {k_bar_l:.3f} bar L")

plt.figure(figsize=(8, 5))
plt.scatter(volume_l, pressure_bar, color="tab:blue", label="Measured pressure")
plt.plot(
    reference_volume_l,
    reference_pressure_bar,
    color="tab:orange",
    label=f"Reference curve (P = k/V, k = {k_bar_l:.3f} bar L)",
)
plt.title("Isothermal Compression of Carbon Dioxide")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.show()
```
